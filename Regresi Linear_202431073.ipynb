# Regresi Linear (Luas Daratan vs Luas Total)

## Langkah 1 - Import library

```python
# Fitrya Irfan_202431073
# Langkah 1: import library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

## Langkah 2 - Baca dataset

```python
# Fitrya Irfan_202431073
# Langkah 2: baca dataset hasil scraping Praktikum 2
df = pd.read_csv("hasil_scraping_negara.csv")
print(df.head())
print("Ukuran dataset:", df.shape)
```

## Langkah 3 - Buang data kosong pada kolom x dan y

```python
# Fitrya Irfan_202431073
# Langkah 3: buang data kosong pada kolom x dan y
df = df.dropna(subset=["Land_Area_km2", "Total_Area_km2"])
print("Jumlah data setelah dropna:", len(df))
```

## Langkah 4 - Tentukan x (fitur) dan y (target)

```python
# Fitrya Irfan_202431073
# Langkah 4: tentukan fitur x dan target y
x = df[["Land_Area_km2"]]     # fitur: luas daratan (km2)
y = df["Total_Area_km2"]      # target: luas total wilayah (km2)
print("Ukuran x:", x.shape)
print("Ukuran y:", y.shape)
```

## Langkah 5 - Bagi data latih dan data uji

```python
# Fitrya Irfan_202431073
# Langkah 5: bagi data latih dan uji (test_size = 0.2, random_state = 0)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
print("Data latih:", len(x_train), "| Data uji:", len(x_test))
```

## Langkah 6 - Buat model LinearRegression

```python
# Fitrya Irfan_202431073
# Langkah 6: buat model regresi linear
model = LinearRegression()
print(model)
```

## Langkah 7 - Latih model

```python
# Fitrya Irfan_202431073
# Langkah 7: latih model dengan data latih
model.fit(x_train, y_train)
print("Model berhasil dilatih")
```

## Langkah 8 - Prediksi data uji

```python
# Fitrya Irfan_202431073
# Langkah 8: prediksi data uji
y_pred = model.predict(x_test)
hasil = pd.DataFrame({"Aktual": y_test.values, "Prediksi": y_pred.round(2)})
print(hasil.head(10))
```

## Langkah 9 - Evaluasi model

```python
# Fitrya Irfan_202431073
# Langkah 9: evaluasi model
print("Koefisien :", model.coef_[0])
print("Intercept :", model.intercept_)
print("MSE       :", mean_squared_error(y_test, y_pred))
print("R2 Score  :", r2_score(y_test, y_pred))
```

## Langkah 10 - Persamaan regresi

```python
# Fitrya Irfan_202431073
# Langkah 10: persamaan regresi Y = a + bx
a = model.intercept_
b = model.coef_[0]
print(f"Y = {a:.4f} + {b:.6f}x")
print("dengan Y = luas total wilayah (km2) dan x = luas daratan (km2)")
```
